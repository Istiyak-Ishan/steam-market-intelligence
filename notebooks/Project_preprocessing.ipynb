```python
import pandas as pd
import numpy as np
import ast
import re
from datetime import datetime
```

```python
df_main = pd.read_csv('steam_games.csv')
df_main.shape
```

```python
df = df_main.copy()
```

```python
column_to_deleted = [
    'short_description', 'about_the_game', 'detailed_description', 
    'notes', 'header_image', 'screenshots', 'movies', 'website', 
    'support_url', 'support_email', 'metacritic_url', 
    'steam_store_available', 'steam_spy_available','packages'
]

df.drop(columns=column_to_deleted, inplace=True, errors='ignore')
df
```

```python
column_to_remove = ['genres', 'categories']

for col in column_to_remove:
    df[col] = df[col].str.replace(r'\{id:\s*\d+,\s*description:\s*', '', regex=True)
    df[col] = df[col].str.replace(r'\}', '', regex=True)
    df[col] = df[col].str.replace(r"['\"]", "", regex=True)
```

```python
df['developers'] = df['developers'].str.replace(r'\[|\]', '', regex=True)
df['publishers'] = df['publishers'].str.replace(r'\[|\]', '', regex=True)
df['genres'] = df['genres'].str.replace(r'\[|\]', '', regex=True)
df['categories'] = df['categories'].str.replace(r'\[|\]', '', regex=True)
```

```python
first_encoder = {True:1,False:0}
df['mac'] = df['mac'].map(first_encoder)
df['windows'] = df['windows'].map(first_encoder)
df['linux'] = df['linux'].map(first_encoder)
```

Null value handeling

As there are 17 Null value in name column,84 null value in release column and 84 price column  we should drop these value

```python
df.dropna(subset=['name'], inplace=True)
df.dropna(subset=['release_date'], inplace=True)
df.dropna(subset=['price'], inplace=True)
df.dropna(subset=['supported_languages'],inplace=True)
```

as there are 19957 Null value in recommendations Column.But it may be true for the game that 0 people recommended these game so we shoud keep those null value and replace it by 0

```python
df['recommendations'] = df['recommendations'].fillna(0)
```

As there  be 0 Metacritic Score so we should keep these value

```python
df['metacritic_score'] = df['metacritic_score'].fillna(0)
```

It may be true to have Zero Achivement in the game so we are keeping these value

```python
df['achievements'] = df['achievements'].fillna(0)
```

We should add none to all the missing value in the full audio language column

```python
df['full_audio_languages'] = df['full_audio_languages'].fillna('None')
```

Now we are creating two new column they lowest estimate owner and high estimate owner

```python
df['estimated_owners'] = df['estimated_owners'].astype(str).str.replace(',', '', regex=False)
split_the_column = df['estimated_owners'].str.split(r'\s*(?:\.+|-+)\s*', expand=True)
df['lowest_estimate_owner'] = pd.to_numeric(split_the_column[0], errors='coerce')
df['highest_estimate_owner'] = pd.to_numeric(split_the_column[1], errors='coerce')
print(df[['lowest_estimate_owner', 'highest_estimate_owner']])
```

```python
df
```

we are creating more two column total revie and review score percentage and has review

```python
df["positive"] = pd.to_numeric(df["positive"], errors="coerce").fillna(0)
df["negative"] = pd.to_numeric(df["negative"], errors="coerce").fillna(0)
df['total_review'] = df['positive']+df['negative']
df["review_score_pct"] = np.where(
        df["total_review"] > 0,
        df["positive"] / df["total_review"],
        np.nan,
    )
df["has_reviews"] = df["total_review"] > 0
df
```

```python
df['primary_genre'] = df['genres'].str.split(',').str[0]
df['genre_count'] = df['genres'].str.split(',').apply(len)
df['categories_count'] = df['categories'].str.split(',').apply(len)
df
```

```python
df['patforms_count'] = df[['mac', 'windows', 'linux']].sum(axis=1)
```

```python
df['languages_count'] = df['supported_languages'].str.split(',').apply(len)
df['primary_language'] = df['supported_languages'].str.split(',').str[0]

df
```

```python
df['age_by_years'] = (datetime.now() - pd.to_datetime(df['release_date'], errors='coerce')).dt.days / 365.25
df['release_quarter'] = pd.to_datetime(df['release_date'], errors='coerce').dt.quarter

df
```

```python
df['has_achievements'] = df['achievements'] > 0
df['achievements_count'] = df['achievements'].apply(lambda x: len(ast.literal_eval(x)) if isinstance(x, str) else 0)
df
```

```python
df['price_tier'] = pd.cut(df['price'], bins=[-1, 0, 10, 30, 60, np.inf], labels=['Free', 'Budget', 'Mid-range', 'Premium', 'AAA'])
df


df
```

```python
df['playtime_engagment_ration'] = df['average_playtime_forever'] / df['median_playtime_forever']+1
df
```

```python
df['playtime_engagment_ratio'] = df['playtime_engagment_ration'].fillna(0)
```

```python
df['recent_vs_lifetime_engagment'] = df['average_playtime_2weeks']/df['average_playtime_forever']+1
df['value_score'] = df['metacritic_score'] / df['price']
```
